Write a Python program to print the current CPU usage percentage.

```python
import psutil

cpu=float(psutil.cpu_percent(interval=1))

print(f"CPU:{cpu}%")
```

Write a program to display total RAM and available RAM.

```python
import psutil
mem=psutil.virtual_memory()
print(f"Memory usage:{mem.percent}%")
print(f"Total Memory:{mem.total/(1024**3):.2f} GB")
print(f"Used Memory:{mem.used/(1024**3):.2f} GB")
print(f"Available Memory:{mem.available/(1024**3):.2f} GB")
```

Write a program to print number of CPU cores.

```python
import psutil
cpu1=psutil.cpu_count()
cpu2=psutil.cpu_count(logical=False)
print(cpu1,"cores")
print(cpu2,"cores")
```

Write a program to display disk usage percentage of root directory.

```python
import psutil

disk=psutil.disk_usage('/')
print(f"disk usage:{disk.percent}%")
print(f"disk Storage:{disk.total/(1024**3):.2f}")
print(f"disk Used:{disk.used/(1024**3):.2f}")
print(f"disk Available:{disk.free/(1024**3):.2f}")
```
